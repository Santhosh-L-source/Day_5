```python
from langchain_core.tools import tool
import sqlite3
```

```python
@tool
def get_student_info(student_id: str) -> str:
    """
    Get the name and department of a student using their student ID.
    Use this tool when the user asks for student name or department.
    """

    conn = sqlite3.connect("students.db")

    cursor = conn.cursor()

    cursor.execute("""
        SELECT name, department
        FROM students
        WHERE student_id = ?
    """, (student_id,))

    result = cursor.fetchone()

    conn.close()

    if result is None:
        return f"No student found with ID {student_id}"

    name, department = result

    return f"Student ID: {student_id}, Name: {name}, Department: {department}"
```

```python
@tool
def get_student_marks(student_id: str) -> str:
    """
    Get the marks of a student in Python, Database, AI and Web.
    Use this tool when the user asks about marks, total, average or passing eligibility.
    """

    conn = sqlite3.connect("students.db")

    cursor = conn.cursor()

    cursor.execute("""
        SELECT python, database_mark, ai, web
        FROM students
        WHERE student_id = ?
    """, (student_id,))

    result = cursor.fetchone()
    conn.close()

    if result is None:
        return f"No student found with ID {student_id}"

    python_mark, database_mark, ai_mark, web_mark = result

    return (
        f"Student ID: {student_id}\n"
        f"Python: {python_mark}\n"
        f"Database: {database_mark}\n"
        f"AI: {ai_mark}\n"
        f"Web: {web_mark}"
    )
```

```python
result_marks = get_student_marks.invoke({"student_id": "22CS045"})
print(result_marks)
```

```python
from typing import Union

@tool
def calculator(expression: str) -> str:
    """
    Calculate a mathematical expression.
    Use this tool for calculating total marks, average marks,
    percentages, or other mathematical calculations.
    """

    try:
        result = eval(expression, {"__builtins__": {}}, {})

        return f"Result: {result}"

    except Exception as e:
        return f"Error calculating expression: {str(e)}"
```

```python
@tool
def get_passing_rules() -> str:
    """
    Get the university passing requirements.
    Use this tool when the user asks whether a student passes
    or satisfies the university passing requirements.
    """

    return """
University Passing Rules:

1. Minimum overall average: 40%
2. Minimum mark in each subject: 35%
"""
```

```python
tools = [
    get_student_info,
    get_student_marks,
    calculator,
    get_passing_rules
]

for tool in tools:
    print(tool.name)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"
)

print("Gemini model created successfully!")

# 1. Get marks
marks = get_student_marks.invoke({
    "student_id": "22CS045"
})

print(marks)


# 2. Calculate total
total = calculator.invoke({
    "expression": "85 + 72 + 90 + 78"
})

print(total)


# 3. Get passing rules
rules = get_passing_rules.invoke({})

print(rules)
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt="""
You are a student information assistant.

You have access to four tools:

1. get_student_info
   - Gets student name and department.

2. get_student_marks
   - Gets Python, Database, AI and Web marks.

3. calculator
   - Performs mathematical calculations such as total and average.

4. get_passing_rules
   - Gets the university passing requirements.

Choose tools based on the user's question.

Do not manually assume a fixed sequence.

If the user asks for total or average:
- First obtain the student's marks.
- Then use the calculator.

If the user asks about passing eligibility:
- Get the student's marks.
- Get the passing rules.
- Calculate the total/average if necessary.
- Check every subject against the minimum subject mark.

If the user asks for name and department:
- Use get_student_info.

If the user asks for marks:
- Use get_student_marks.

Always provide a clear final answer based only on the
information returned by the tools.
"""
)

print("Agent created successfully!")
```

```python
def ask_agent(question):
    response = agent.invoke({
        "messages": [
            {
                "role": "user",
                "content": question
            }
        ]
    })
    return response
```

```python
response = ask_agent(
    "What is the name and department of student 22CS045?"
)

print(response["messages"][-1].content)
```
